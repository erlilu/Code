```python
from matplotlib import axis

"""
题目 1：温度数据分析
某城市一周的最高气温为[28, 30, 29, 31, 32, 30, 29]。
- 计算平均气温、最高气温和最低气温。
- 找出气温超过30摄氏度的天数。
"""
```

```python
import numpy as np
temps = np.array([28, 30, 29, 31, 32, 30, 29])
print( f'平均气温为: ','%.2f'%np.mean(temps))
print(f'最高气温为:{np.max(temps)}，最低气温为：{np.min(temps)}')
print('气温超过30度的天数：',temps[temps > 30].size)
print('气温低于30度的天数:', np.where(temps < 30)[0].size)
print('气温等于30度的天数：', np.cumsum(np.where(temps == 30, 1, 0))[-1])
print('气温等于30度的天数：', np.count_nonzero(temps == 30))
```

```python
"""
题目 2：学生成绩统计
某班级 5 名学生的数学成绩为 [85, 90, 78, 92, 88]
- 计算成绩的平均分、中位数和标准差。
- 将成绩转换为（假设满分为10）
"""
```

```python
score = np.array([85, 90, 78, 92, 88])
print('平均分：',np.mean(score), '中位数：', np.median(score), '标准差：', '%.3f'%np.std(score))
print(score/10)
```

```python
"""
题目 3：矩阵运算
给定矩阵 A = [[1, 2], [3, 4]]  和 B = [[5, 6], [7, 8]]
- 计算 A + B 和 A * B（逐元素乘法）
- 计算 A 和 B 的矩阵乘法（点积）
"""
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print('A + B = ', A + B)
print('A * B = ', A * B)
print('A 和 B的矩阵乘法：', np.dot(A, B))
print('A 和 B的矩阵乘法：', A @ B)
```

```python
"""
题目 4： 随机数据生成
生成一个(3， 4)的随机整数数组，范围[0， 10)。
- 计算每列的最大值和每行的最小值。
- 将数组的所有奇数替换为 -1.
"""
```

```python
np.random.seed(0)
arr = np.random.randint(0,10,(3,4))
print(arr)
print('列最小值：', np.min(arr, axis=0), '行最小值：', np.min(arr, axis=1))       # axis=0为列，=1为行
print('列最大值：', np.max(arr, axis=0), '列最大值：', np.max(arr, axis=1))
print(np.where(arr % 2 == 1, -1, arr))
arr[arr % 2 == 1] = -1
print(arr)
```

```python
"""
题目 5: 数组变形
创建一个 1 到 12 的一维数组，并转换为(3, 4)的二维数组。
- 计算每行的和 与 每列的平均值。
- 将数组展平为一维数组。
"""
```

```python
arr1 = np.arange(1, 13)
print(arr1)
arr2 = np.reshape(arr1, (3, 4))
print(arr2)
print('每行的和：', np.sum(arr2, axis=1))
print('每列的平均值：', np.median(arr2, axis=0))
print(np.reshape(arr2, (12)))
```

```python
"""
题目 6：布尔索引
生成一个(5， 5)的随机数组，范围[0， 20)。
- 找出数组中大于 10 的元素。
- 将所有大于 10 的元素替换为 0。
"""
```

```python
np.random.seed(0)
arr = np.random.randint(0, 20, (5, 5))
print(arr)
print(arr[arr > 10])
arr[arr > 10] = 0
print(arr)
```

```python
"""
题目 7: 统计函数应用
某公司 6 个月的销售额(万元) 为 [120, 135, 110, 125, 130, 140]。
- 计算销售额的总和、均值和方差。
- 找出销售额最高的月份和最低的月份。
"""
```

```python
money = np.array([120, 135, 110, 125, 130, 140])
print('销售额的总和：', np.sum(money), '均值：', '%.3f'%np.median(money), '方差：', '%.3f'%np.var(money))
print('销售最高的月份：', np.argmax(money), '最低的月份：', np.argmin(money))
```

```python
"""
题目 8: 数组拼接
给定 A = [1, 2, 3] 和 B = [4, 5, 6]。
- 将 A 和 B 水平拼接为一个新数组。
- 将 A 和 B 垂直拼接为一个新数组。
"""
```

```python
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
print(np.concatenate((A, B)))
print(np.reshape(np.concatenate((A, B)), (2, 3)))
```

```python
"""
题目 9： 唯一值与排序
给定数组 [2, 1, 2, 3, 1, 4, 3]
- 找出数组中的唯一并排序。
- 计算每个唯一值出现的次数。
"""
```

```python
arr = np.array([2, 1, 2, 3, 1, 4, 3])
uni_arr, counts = np.unique(arr, return_counts=True)
print(uni_arr)
print(counts)
```

```python
"""
题目 10: 综合应用
某商店 5 天的销售额（万元）和成本（万元）如下：
销售额：[20, 25, 22, 30, 28]
成本：[15, 18, 16, 22, 20]
- 计算每天的利润（销售额 - 成本）。
- 计算利润的平均值和标准差。
- 找出利润最高的天数。
"""
```

```python
money = np.array([20, 25, 22, 30, 28])
base = np.array([15, 18, 16, 22, 20])
earn = money - base
print('每天的利润',earn)
print('利润的平均值：', np.mean(earn), '利润的标准差：', '%.3f'%np.std(earn))
print('利润最高的天数：', len(earn[earn == np.max(earn)]))
```
