```python
from operator import index

import numpy as np
import pandas as pd
```

# series的创建
1. 直接创建，自定义index和name
2. 字典创建

```python
# series的创建
s = pd.Series([10,2,3,4,5])

# 自定义索引
s = pd.Series([10,2,3,4,5], index=['a','b','c','d','e'])

# 定义name
s = pd.Series([10,2,3,4,5], index=['a','b','c','d','e'], name='month')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({"a": 1, "b": 2, "c": 3, "d": 4, "e": 5}, name="test")
# print(s)

s1 = pd.Series(s, index=['a', 'c'])
print(s1)
```

# Series的属性
1. index: Series的索引对象。
2. values: 数值。
3. dtype/dtypes: 元素类型。
4. shape: 形状。
5. ndim: 维度。
6. size: 元素的个数。
7. name: 名称。
8. loc[]: 显示索引，按标签索引或切片。
9. iloc[]: 隐式索引，按位置索引或切片。
10. at[]: 使用标签访问单个元素。
11. iat[]: 使用位置访问单个元素。

```python
print(s.index)
print(s.values)
print(s.shape, s.ndim, s.size)
print(s.dtype, s.name)
print(s.loc['b' : 'd'])   # 显示索引，支持切片
print(s.iloc[1])    # 隐式索引，支持切片【默认下标】
print(s.at['a'])    # 同显式，但不支持切片
print(s.iat[2])   # 同隐式，但不支持切片
```

## 访问数据
1. []
2. head()
3. tail()

```python
print(s[1])
print(s['c'])
print(s[s < 3])
print(s.head(2))     # 默认前5行
print(s.tail(1))     # 默认后5行
```

# Series的常用方法
1. head()/tail(): 查看前/后几行的数据，默认5行。
2. isin(): 判断元素是否在参数集合中。
3. isna(): 判断是否为缺失值(如NaN或None)。
4. sum()/mean(): 求和/求平均值。
5. min()/max(): 最小值/最大值。
6. idxmax()/idxmin(): 最大值/最小值的下标。
7. var()/std(): 方差差/标准差。
8. median(): 中位数。【不包含NaN】
9. mode(): 众数(可返回多个)。
10. quantile(q): 分位数，q取0-1之间。
11. describe(): 常见统计信息(count, mean, std, min, 25%, 50%, 75%, max)
12. value_counts():每个唯一值的出现次数。
13. count(): 非缺失值的数量。
14. nunique(): 唯一值的个数。
15. unique(): 获取去重后的值数组。
16. drop_duplicates(): 去除重复项。
17. sample(): 随机抽样。
18. sort_index(): 按索引排序。
19. sort_values(): 按值排序。
20. replace(): 替换值。
21. keys()/index: 返回Series的索引对象。
22. diff(): 计算Series得变化值。
23. pct_change(): percent百分比的变化。
24. resample()：重新采样。‘QS’：按季度。
25. rolling(): 滑动窗口。
26. between_time('start_time', 'end_time'):时间范围。
27. drop(): 去除需要去除的数据。
28. nlargest():数值最大的数，可以设置多少个数。

```python
# 常见函数
s = pd.Series([10, 2, np.nan, None, 3, 4, 5], index=['a','b','c','d','e','f', 'g'], name = 'data')
print(s)
```

```python
print(s.describe())
```

```python
print(s.median())
print(s.quantile(0.75))
print(s.quantile(0.8))
```

```python
s['H'] = 4
print(s.mode())
print(s.value_counts())
print(s.drop_duplicates())
```
