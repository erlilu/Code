```python
import pandas as pd
import numpy as np
```

# DateFrame的创建
1. 通过Series创建
2. 通过字典来创建

```python
# 通过Series创建
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
df = pd.DataFrame({"第1列：": s1, "第2列：": s2})
df
```

```python
# 通过字典来创建
df = pd.DataFrame(
    {
        "id": [1,2,3,4,5],
        "name": ['tom', 'jack', 'alice', 'bob', 'allen'],
        "age": [17, 17, 20, 26, 30],
        "score": [60.5, 80, 30.6, 70, 83.5]
    }, index = [1, 2, 3, 4, 5], columns = ["name", "score", "age"]
)
df
```

# DataFrame属性
1. index: 行索引
2. values:值
3. dtypes:元素类型
4. shape:形状
5. ndim:维度
6. size:元素个数
7. columns:列标签
8. loc[]:显示索引，按行列标签索引或切片
9. iloc[]:隐式索引，按行列位置索引或切片【默认下标】
10. at[]:使用行列标签访问单个元素
11. iat[]:使用行列位置访问单个元素
12. T:转置。

```python
print("行索引：", df.index)
print("列标签：", df.columns)
print("值：", df.values)
print('维度：', df.ndim)
print('形状', df.shape)
print('元素个数：',df.size)
print("数据类型：\n", df.dtypes)
```

```python
print(df.T)
```

```python
# 获取元素
print(df.loc[2 : 4])
print('-----------------')
print(df.iloc[0])
```

```python
# 某列
print(df.loc[:, 'name'])
print(df.iloc[:, 0])
```

```python
# 单个元素
print(df.at[1, 'age'])
print(df.iat[3, 1])
```

```python
# 获取单列数据
print(df['name'], type(df['name']))
print(df[['name']], type(df[['name']]))
```

```python
# 查看部分数据
print(df.head(3))
print(df.tail(3))
```

```python
df[df.score > 60]   # bool索引
```

# DateFrame的常用方法【同Series】
1. head()/tail(): 查看前/后几行的数据，默认5行。
2. isin(): 判断元素是否在参数集合中。
3. isna(): 判断是否为缺失值(如NaN或None)。
4. sum()/mean(): 求和/求平均值。
5. min()/max(): 最小值/最大值。
6. idxmax()/idxmin(): 最大值/最小值的下标。
7. var()/std(): 方差差/标准差。
8. median(): 中位数。【不包含NaN】
9. mode(): 众数(可返回多个)。
10. quantile(q): 分位数，q取0-1之间。
11. describe(): 常见统计信息(count, mean, std, min, 25%, 50%, 75%, max)
12. value_counts():每个唯一值的出现次数。
13. count(): 非缺失值的数量。
14. duplicated(): 是否重复。
15. sample(): 随机抽样。
16. sort_index(): 按索引排序。
17. sort_values(): 按值排序。
18. replace(): 替换值。
19. nlargest()： 返回某列最大的n条数据。
20. nsmallest()：返回某列最小的n条数据。

```python
print(df['score'].sum())
df.sum()
```

```python
print(df.score.max())
print(df.name.min())
print(df.score.mean())
print(df.score.median())
print(df.score.var())
print(df.score.std())
print(df.age.mode())
```
