# axis
正确的理解方式：axis 指的是“沿着哪个轴移动/压缩”，而不是“结果留在哪”。
> 看到 axis，就在心里默念：0 是行（竖着干），1 是列（横着干）。

- axis=0 是垂直向下移动。你从上往下扫过所有行，把行压扁了，所以它针对的是行。
- axis=1 是水平向右移动。你从左往右扫过所有列，把列压扁了，所以它针对的是列。

# [[]]
> df['列名'] 是把这列“抽”出来（变细了），df[['列名']] 是把这个表格“裁”出一列（依然是表格）。
- 在你的代码中，如果你只是想知道平均值，用单括号 df['wind'].mean() 更简洁；如果你想把结果赋给另一个变量并保留列名，就用双括号。

## 缺失值处理
1. 查看缺失值【是否有缺失值、查看缺失值个数】
2. 剔除缺失值
3. 缺失值填充【字典填充、统计值填充、前后填充】

```python
# 缺失值的处理
# nan = not a number

s = pd.Series([1, 2, np.nan, None, pd.NA])
df = pd.DataFrame([[1, pd.NA, 2], [2, 3, 4], [None, 4, 6]], columns=['a', 'b', 'c'])
print(s.isna())
print(s.isnull())
print(df.isna())
print(df.isnull())
print(df.isna().sum(axis = 1))  # 查看缺失值的个数
print(s.isna().sum())   # 查看缺失值的总数

# 剔除缺失值
print(s.dropna())
print(df.dropna())      # 含有na数据的内容，整条数据全部删除
print(df.dropna(how='all')) # 如果所有的值都是缺失值，删除这一行
print(df.dropna(thresh=2))          # 如果至少有n个值不是缺失值，就保留
print(df.dropna(axis=1,))           # axis = 1 作用于列
df.dropna(subset=['a'])             # 如果某列有缺失值，则删除这一行
```

```python
# 填充缺失值【字典填充、统计值填充、前后填充】
df = pd.read_csv('data/weather_withna.csv')
print(df.isna().sum(axis = 0))
df.head()
# 字典填充
df.fillna({'temp_max': 20, 'wind': 2.5})     # 将temp_max为na的值改为默认20
# 使用统计值来填充
df.fillna(df[['temp_max', 'wind']].mean()).tail()           # 【用平均值来填充】
# 前后填充
df.ffill().tail()           # front使用前面的来填充
df.bfill().tail()           # back使用后面的来填充
```

## 数据分箱
### 等份切分
1. pd.cut(x, bins, labels)
2. x：要处理的一维数组、Series 或 DataFrame 列。
3. bins：分段规则，有两种写法：
    - 整数：表示等宽分箱的数量（如 bins=4 会把数据按值域等分成4段）。
    - 序列（列表/数组）：自定义分割点（如 [0, 18, 35, 60, 100] 会生成 [0,18], [18,35], [35,60], [60,100] 四个区间）。
4. labels（可选）：给每个区间起一个名字（如 ['少年', '青年', '中年', '老年']）。如果不提供，默认返回的是区间范围字符串（如 (18, 35]）。
------
- 如果你只是想简单地把年龄、收入、评分等连续变量转换成分类变量，pd.cut() 就是你的首选工具。
- 如果你想给区间起名字，记得用 labels 参数，这样后续分析和可视化会清晰很多。
- 如果数据有边界值问题，检查一下 right 和 include_lowest 参数是否设置正确。【左开右闭】
### 等频切分
1. pd.qcut(x, q, labels)
- x：一维数组或 Series。
- q：整数（如 q=4 表示四分位数，q=10 表示十分位数）或分位数列表（如 [0, 0.25, 0.5, 0.75, 1]，和 q=4 等价）。
- labels：和 pd.cut() 一样，可以给区间起名字。如果不传，默认返回区间范围（如 (32.5, 57.5]）。

```python
df = pd.read_csv('data/employees.csv')
df1 = df.head(10)[['employee_id', 'salary']]
df1
```

```python
pd.cut(df1['salary'], bins= 3)
pd.cut(df1['salary'], bins= 3).value_counts()
pd.cut(df1['salary'], bins= [0, 10000, 20000, 30000])
pd.cut(df1['salary'], bins= [0, 10000, 20000, 30000]).value_counts()
pd.cut(df1['salary'], bins= [0, 10000, 20000, 30000], labels=['low', 'mid', 'high']).value_counts()
```

```python
# 等频拆分
pd.qcut(df1['salary'], 3).value_counts()
```

## 分组聚合
> 先分组，再聚合
1. groupby('分组的字段')['聚合的字段'].聚合函数()

```python
df = pd.read_csv('data/employees.csv')
print(df['department_id'].isna().sum())     # 只有一行，可以直接删掉
df.dropna(subset=['department_id'], inplace=True)
df['department_id'] = df['department_id'].astype('int64')
df.info()
```

```python
# 计算不同部门的平均薪资
# 分组
df.groupby('department_id').groups      # 查看分组
df.groupby('department_id').get_group(10)   # 查看具体的某个分组数据
# 分组聚合
df2 = df.groupby('department_id')['salary'].mean().round(2)       # round()保留几位小数
df2.reset_index()
```

```python
# 计算不同部门、不同岗位的人的平均薪资
df.groupby(['department_id', 'job_id']).groups
df.groupby(['department_id', 'job_id'])['salary'].mean()
```
