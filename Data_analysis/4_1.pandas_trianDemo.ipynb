```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/penguins.csv')
df.head()
df.info()
```

```python
# 数据清洗
df.isna().sum()
df.dropna(inplace=True)     # 直接剔除NA值
df.isna().sum()
```

```python
# 数据特征构造
df.head()
df['sex'] = df['sex'].astype('category')        # sex设为category
df['bill_ratio'] = df['bill_length_mm'] / df['bill_depth_mm']
df.info()
```

```python
# 数据分析
# 分箱 --- 体重氛围三份(low, mid, high)
labels = ['thin', 'mid', 'heavy']
df['mass_level'] = pd.cut(df['body_mass_g'], bins= 3, labels=labels )
print(df['mass_level'].value_counts())

# 按岛屿、性别分组
df.groupby(['island', 'sex']).agg({
    'body_mass_g': ['mean', 'count'],
})      # 数据量大，用agg来处理
```
