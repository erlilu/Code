```python
# 绘制折线图
import matplotlib.pyplot as plt         # 图表包
from matplotlib import rcParams         # 设置字体
rcParams['font.family'] = 'STHeiti'     # mac   win:SimHei
# rcParams['font.sans-serif'] = 'STHeiti'
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 增加数据
month = ['1月', '2月', '3月', '4月']
sales = [100, 150, 80, 130]

# 绘制图表 【先x后y】
plt.plot(month, sales,
         label='产品A',
         color='blue',
         linewidth=2,
         linestyle='-',
         marker='o',)

# 添加图表配置
plt.title('2025销售趋势', color='red', fontsize=20)     # 添加标题
# 设置标题、字体大小
plt.xlabel('月份', fontsize=10)                       # x轴标题
plt.ylabel('销售额（万元）', fontsize=10)                  # y轴标题
# 设置刻度字体大小【旋转角度，字体大小】
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.legend(loc='upper left', fontsize=10)               # 图例【绘制时，配置】
plt.grid(True, alpha=0.1, color='white', linestyle='--')        # 网格线：True：xy轴都有；axis=‘x',axis='y'，选择xy轴的网格
# 设置xy轴的范围
plt.ylim(0, 200)
# 在每个数据点上显示数据
for x,y in zip(month, sales):
    plt.text(x, y, str(y), ha='center', va='bottom', fontsize=10)   # x,y可以理解为坐标、ha水平居中、va垂直居中、字体大小
# 显示图表
plt.show()
```

```python
# 绘制条形图
import matplotlib.pyplot as plt         # 图表包
from matplotlib import rcParams         # 设置字体
rcParams['font.family'] = 'STHeiti'     # mac   win:SimHei
# rcParams['font.sans-serif'] = 'STHeiti'
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 增加数据
subject = ['语文', '数学', '英语', '科学']
score = [85, 92, 78, 90]

# 绘制柱状图【bar竖柱状图，barh横柱状图】
plt.bar(subject, score,
         label='分数',
         color='blue',
         width=0.5)

# 添加图表配置
plt.title('2025成绩分布', color='red', fontsize=20)     # 添加标题
# 设置标题、字体大小
plt.xlabel('科目', fontsize=10)                  # x轴标题
plt.ylabel('分数', fontsize=10)                  # y轴标题
# 设置刻度字体大小【旋转角度，字体大小】
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=10)

plt.legend(loc='upper left', fontsize=10)               # 图例【绘制时，配置】
plt.grid(axis='y', alpha=0.1, color='white', linestyle='--')        # 网格线：True：xy轴都有；axis=‘x',axis='y'，选择xy轴的网格
# 设置xy轴的范围
plt.ylim(0, 100)
# 在每个数据点上显示数据
for x,y in zip(subject, score):
    plt.text(x, y, str(y), ha='center', va='bottom', fontsize=10)   # x,y可以理解为坐标、ha水平居中、va垂直居中、字体大小

# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制条形图
import matplotlib.pyplot as plt         # 图表包
from matplotlib import rcParams         # 设置字体
rcParams['font.family'] = 'STHeiti'     # mac   win:SimHei
# rcParams['font.sans-serif'] = 'STHeiti'
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 增加数据
countries = ['United States', 'China', 'Japan', 'Germany', 'India']
gdp = [92, 78, 43, 22, 8]

# 绘制柱状图【bar竖柱状图，barh横柱状图】
plt.barh(countries, gdp,
         height=0.5)
# 设置标题、字体大小
plt.xlabel('gdp', fontsize=10)                  # x轴标题
plt.ylabel('countries', fontsize=10)                  # y轴标题

# 添加图表配置
plt.title('GDP', color='red', fontsize=20)     # 添加标题

# 设置刻度字体大小【旋转角度，字体大小】
plt.xticks(rotation=0, fontsize=10)
plt.yticks(rotation=0, fontsize=10)

# 自动优化排版
plt.tight_layout()
# 显示图表
plt.show()
```

```python
# 绘制饼状图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

# 设置图标大小
plt.figure(figsize=(5,5))

# 创建数据
things = ['study', 'sleep', 'sport', 'entertainment', 'other']
times = [4, 8, 1, 5, 6]
# colors = ['red', 'green', 'blue', 'yellow', 'orange']   # 不设置即为默认
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']   # 不设置即为默认
# 绘制饼图
plt.pie(times, labels=things,
        autopct='%.1f%%',
        startangle=90,
        colors=colors)   # 需要切分的数据放在前面,'%%'是格式界定符,startangle初始角度
# 添加标签
plt.title('times of one day', color='red', fontsize=20)
plt.tight_layout()
plt.show()
```

```python
# 绘制环形图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

plt.figure(figsize=(10,5))

things = ['study', 'sleep', 'sport', 'entertainment', 'other']
times = [4, 8, 1, 5, 6]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']

plt.pie(times, labels=things,
        autopct='%.1f%%',
        startangle=90,
        colors=colors,
        wedgeprops={'width': 0.3 },
        pctdistance=0.8)     # wedgeprops:设置环；width--设置圆环的宽度；ctdistance--设置内百分比位置

plt.title('times of one day', color='red', fontsize=20)
plt.text(0,0, 'summary:\n100%', ha='center', va='center', fontsize=20)
plt.tight_layout()
plt.show()
```

```python
# 绘制爆炸式饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

plt.figure(figsize=(10,5))

things = ['study', 'sleep', 'sport', 'entertainment', 'other']
times = [4, 8, 1, 5, 6]
colors = ['#66b3ff', '#99ff99', '#ffcc99', '#ff9999', '#ff4499']
explode = [0, 0.1, 0, 0, 0]         # 设置突出块的位置

plt.pie(times, labels=things,
        autopct='%.1f%%',
        startangle=-90,
        colors=colors,
        explode=explode,    # 设置突出块
        shadow=True)        # 设置阴影

plt.title('times of one day', color='red', fontsize=20)
plt.tight_layout()
plt.show()
```

```python
import random
# 绘制散点图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

plt.figure(figsize=(10,8))
x = []
y = []
for i in range(1000):
    x.append(random.uniform(0, 10))
    y.append(2*x[i] + random.gauss(0,3))

plt.scatter(x,y,
            color='blue',
            alpha=0.5,)         # 设置透明度

plt.plot([0,10], [0,20], color='red', linewidth=2)
plt.show()
```

```python
# 绘制箱线图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

plt.figure(figsize=(10,8))

data = {
    'chinese': [82, 85, 88, 70, 90, 76, 84, 83, 95],
    'math' : [75, 80, 79, 80, 45, 65, 88, 35, 92],
    'english' : [70, 78, 59, 68, 90, 46, 88, 45, 90]
}

plt.boxplot(data.values(), tick_labels=data.keys(),)

plt.title('score of subjects(boxplot)')
plt.grid(axis='y', alpha=0.5, color='white', linestyle='--')

plt.show()
```

```python
# 多个图的绘制
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'STHeiti'

month = ['1', '2', '3', '4', '5', '6', '7', '8', '9']
sales = [100, 150, 80, 130, 68, 156, 68, 99, 140]

f1 = plt.subplot(2,2,1)      # 创建子图---2行2列，序号1
f1.plot(month, sales, color='blue', linewidth=2)
f2 = plt.subplot(2,2,2)
f2.bar(month, sales, color='pink', linewidth=2)
f3 = plt.subplot(2,2,3)
f3.scatter(month, sales, color='orange', linewidth=2)
f4 = plt.subplot(2,2,4)
f4.barh(month, sales, color='green', linewidth=2)

```
