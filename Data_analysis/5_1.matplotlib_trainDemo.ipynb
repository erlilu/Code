```python
"""
分析案例： 温度分析
1. 导入库
2. 导入数据
3. 绘图
    3.1 绘制气温的趋势变化图
    3.2 绘制降水量的直方图
"""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = ['STHeiti']

df = pd.read_csv('data/weather.csv')
print(df.head())

df['date'] = pd.to_datetime(df['date'])
df = df[df['date'].dt.year == 2015]

plt.figure(figsize=(20, 10))
plt.plot(df['date'], df['temp_max'], label='max_temp')
plt.plot(df['date'], df['temp_min'], label='min_temp')
plt.title('Chart showing temperature trends of 2015 ')
plt.xlabel('date')
plt.ylabel('temperature')
plt.legend()
```

```python
# 平均气温
df['temp_mean'] = (df['temp_max'] + df['temp_min'])/2
df.head()

plt.figure(figsize=(20, 10))
plt.plot(df['date'], df['temp_mean'], label='temp_mean')
plt.xlabel('date')
plt.ylabel('temperature')
plt.legend()
plt.show()
```

```python
# 绘制降水量的直方图
plt.hist(df['precipitation'], bins=5)
```
