```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['STHeiti']

penguins = pd.read_csv('data/penguins.csv')
penguins.dropna(inplace=True)
penguins.head()
```

```python
# 直方图
sns.histplot(data=penguins, x=penguins['species'])
```

# 核密度估计图
1. 核密度估计图(KDE， Kernel Density Estimate Plot)是一种用于显示数据分布的统计图表，它通过平滑直方图的方法估计数据的概率密度函数，使得分布图看起来更加平滑。
2. 核密度估计是一种非参数方法，且用于估计随机变量的概率密度函数。
3. 基本思想：将每个数据点视为一个“核”(通常是高斯分布)，然后将这些核的贡献相加以形成平滑的密度曲线。

# 计数图
计数图用于绘制分类变量的技术分布图，显示每个类别在数据集中出现的次数，是分析分类数据非常直观的工具，可以快速了解类别的分布情况。

kdeplot()，同时设置x参数和y参数来绘制二维核密度估计图。

```python
sns.histplot(data=penguins, x='bill_length_mm', kde=True)
```

```python
# 绘制不同岛屿企鹅数量的计数图【计数图和直方图做统计时候类似】
sns.countplot(data=penguins, x='island')      # 计数图
sns.histplot(data=penguins, x='island')       # 直方图
```

```python
# 散点图
# 绘制横轴为体重，纵轴为脚蹼长度的散点图。 可通过hue参数设置不同组别进行对比。
sns.scatterplot(data=penguins, x='body_mass_g', y='flipper_length_mm', hue='sex')
```

```python
# 蜂窝图
sns.jointplot(data=penguins, x ='body_mass_g', y = 'flipper_length_mm', kind='hex')
```

```python
# 二维核密度估计图
sns.kdeplot(data=penguins, x='body_mass_g', y='flipper_length_mm')
# sns.kdeplot(data=penguins, x=penguins['body_mass_g'], y=penguins['flipper_length_mm'], shade=True)
```

```python
# 通过fill=True设置为填充，通过cbar=True设置显示颜色示意图
sns.kdeplot(data=penguins, x='body_mass_g', y='flipper_length_mm', fill=True, edgecolor='black', cbar=True)
```

```python
# 条形图
sns.barplot(data=penguins, x='species', y='bill_length_mm', estimator='mean', errorbar=None)
```

```python
# 箱线图
sns.boxplot(data=penguins, x='species', y='bill_length_mm')
```

# 小提琴图 Violin Plot
1. 小提琴图是一种结合了箱线图和核密度估计图KDE的可视化图表，用于展示数据的分布情况、集中趋势、散布情况以及异常值。
2. 显示数据的基本统计量（如中位数、四分位数）
3. 展示数据的概率密度，提供比箱线图更丰富的信息

# 成对关系图
1. 成对关系图是一种用于显示多个变量之间关系的可视化工具。
2. 展示各个变量之间的成对关系，并且通过不同的图表形式帮助我们理解数据中各个变量之间的相互作用。
3. 对角线上的图通常显示每个变量的分布(如直方图或核密度估计图)，帮助观察每个变量的单变量特性。
4. 其他位置展示所有变量的两两关系，用散点图表示。

```python
# violin plot
sns.violinplot(data=penguins, x='species', y='bill_length_mm')
```

```python
# 成对关系图
sns.pairplot(data=penguins, hue='species')
```
