梯度grad

```python
w = 0.0                    # 初始参数
lr = 0.1                   # 学习率
y_true = 2.0               # 真实值
x = 1.0                    # 输入

for step in range(20):
    y_pred = w * x                          # 前向传播：预测
    loss = (y_true - y_pred) ** 2            # 计算损失
    grad = -2 * (y_true - y_pred)            # 计算梯度 dL/dw
    w = w - lr * grad                        # 梯度下降：更新参数
    print(f"step {step}, w={w:.4f}, loss={loss:.4f}")
```
