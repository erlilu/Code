```python
import numpy as np
import matplotlib.pyplot as plt
```

设$f(x)=x^2$,求x为何值时，$f(x)=2.$

```python
# 目标函数
def J(x):
    return (x ** 2 - 2) ** 2
# 梯度函数
def gradient(x):
    return 4 * x ** 3 - 8 * x
```

```python
# 记录x，y
x_list = []
y_list = []
```

```python
# 定义x的初始值、超参数
x = 1
alpha = 0.1
```

```python
# 重复迭代100次
while np.abs(gradient(x)) > 1e-10:
    y = J(x)
    x_list.append(x)
    y_list.append(y)
    print(f'x={x}, y={y}')
    # 计算梯度
    grad = gradient(x)
    # 更新参数
    x -= alpha * grad
print(len(x_list))
```

```python
# 可视化
x = np.arange(0.9, 1.6, 0.01)
plt.plot(x, J(x), color='green')
plt.plot(x_list, y_list, color='red')
plt.scatter(x_list, y_list, color='blue')
plt.show()
```

```python
# 局部放大
fig, ax = plt.subplots(1, 2, figsize=(15, 4))
x = np.arange(0.9, 1.6, 0.01)
ax[0].plot(x, J(x), color='green')
ax[0].plot(x_list, y_list, color='red')
ax[0].scatter(x_list, y_list, color='blue')
ax[0].grid(True)

x_list2 = x_list[1:]
y_list2 = y_list[1:]
x = np.arange(1.395, 1.42, 0.001)
ax[1].plot(x, J(x), color='green')
ax[1].plot(x_list2, y_list2, color='red')
ax[1].scatter(x_list2, y_list2, color='blue')
ax[1].grid(True)

plt.show()
```

改变超参数大小，alpha = 0.01
> 训练次数增加

```python
# 目标函数
def J(x):
    return (x ** 2 - 2) ** 2
# 梯度函数
def gradient(x):
    return 4 * x ** 3 - 8 * x

x_list = []
y_list = []

x = 1
alpha = 0.01

# 重复迭代
while np.abs(gradient(x)) > 1e-10:
    y = J(x)
    x_list.append(x)
    y_list.append(y)
    print(f'x={x}, y={y}')
    # 计算梯度
    grad = gradient(x)
    # 更新参数
    x -= alpha * grad
print(len(x_list))


# 局部放大
fig, ax = plt.subplots(1, 2, figsize=(15, 4))
x = np.arange(0.9, 1.6, 0.01)
ax[0].plot(x, J(x), color='green')
ax[0].plot(x_list, y_list, color='red')
ax[0].scatter(x_list, y_list, color='blue')
ax[0].grid(True)

x_list2 = x_list[1:]
y_list2 = y_list[1:]
x = np.arange(1.395, 1.42, 0.001)
ax[1].plot(x, J(x), color='green')
ax[1].plot(x_list2, y_list2, color='red')
ax[1].scatter(x_list2, y_list2, color='blue')
ax[1].grid(True)

plt.show()
```

改变超参数大小，alpha = 0.2【只能手动确定次数】
> 过大，直接发散，无法准确收敛；

```python
# 目标函数
def J(x):
    return (x ** 2 - 2) ** 2
# 梯度函数
def gradient(x):
    return 4 * x ** 3 - 8 * x

x_list = []
y_list = []

x = 1
alpha = 0.2

# 重复迭代100次
for i in range(100):
    y = J(x)
    x_list.append(x)
    y_list.append(y)
    print(f'x={x}, y={y}')
    # 计算梯度
    grad = gradient(x)
    # 更新参数
    x -= alpha * grad
print(len(x_list))

# 可视化
plt.plot(x, J(x), color='green')
plt.plot(x_list, y_list, color='red')
plt.scatter(x_list, y_list, color='blue')
plt.show()

plt.show()
```
