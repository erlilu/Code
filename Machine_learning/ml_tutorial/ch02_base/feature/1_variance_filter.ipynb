```python
import numpy as np
```

```python
# 构造特征
a = np.random.randn(100)    # 服从标准正态分布
b = np.random.randn(100) * 0.1      # *0.1后，方差直接变为0.01
c  = np.random.normal(5, 0.1, 100)  # (均值，标准差，数量)
print(np.var(a))
print(np.var(b))
print(np.var(c))
```

```python
# 构造特征向量（输入数据X）
X = np.vstack((a, c)).T       # vstack创建竖直的两组随机数,转置变为需要的（n，2）
X.shape
```

- X_filtered = vt.fit_transform(X)：对原始特征矩阵 X 执行“拟合+变换”。它会自动计算每个特征的方差，把方差低于 0.01 的列删掉，返回过滤后的新特征矩阵。
- print(X_filtered.shape)：打印新矩阵的形状 (样本数, 剩余特征数)，让你直观看到删掉了多少列。

```python
# 低方差过滤
from sklearn.feature_selection import VarianceThreshold
vt = VarianceThreshold(0.01)        # 移除所有方差小于 0.01 的特征
X_filtered = vt.fit_transform(X)
print(X_filtered)
print(X_filtered.shape)
```
