```python
import pandas as pd
```

```python
# 读取数据
advertising = pd.read_csv('../../data/advertising.csv')
print(advertising.head())
print(advertising.describe())
print(advertising.shape)
```

```python
# 数据预处理
# 去掉第一列，无意义的数
advertising.drop(columns='Unnamed: 0', inplace=True)
# 去掉空值
advertising.dropna(inplace=True)
# 提取特征和标签(目标值)
X = advertising.drop(columns='Sales')
y = advertising['Sales']
```

```python
print(X.shape)
print(y.shape)
```

# 相关系数
- corrwith() 相关系数--和其他变量的相关系数
- corr() 直接算x内部的相关系数

```python
# 计算相关系数（使用皮尔逊相关系数）
X.corrwith(y, method='pearson')
```

```python
corr_matrix = advertising.corr(method='pearson')
```

```python
# 将相关系数矩阵画成热力图
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(corr_matrix, annot=True, cmap='YlGnBu', fmt='.2f')    # annot 确认是否显示数值
plt.show()
```
