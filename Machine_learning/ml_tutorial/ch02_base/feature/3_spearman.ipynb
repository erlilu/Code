```python
import pandas as pd
```

```python
# 定义数据
X = [[5], [8], [10], [12], [15], [3], [7], [9], [14], [6]]
y = [55, 65, 70, 75, 85, 50, 60, 72, 80, 58]
```

```python
X = pd.DataFrame(X)
y = pd.Series(y)
X.shape, y.shape
```

```python
# 调用 spearman 相关系数
res = X.corrwith(y, method="spearman")
res
```
