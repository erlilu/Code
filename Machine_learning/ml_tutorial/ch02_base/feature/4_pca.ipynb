```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
# 生成数据
X = np.random.randn(1000, 3)
```

```python
# 使用PCA进行降维，将三维数据降成2维
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
print(X_pca.shape)
```

```python
# 可视化
# 原始数据3维图
fig = plt.figure(figsize=(12, 4))
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(X[:, 0], X[:, 1], X[:, 2], c='g')
ax1.set_title('Before PCA')
ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')
ax1.set_zlabel('Feature 3')

# PCA降维后的2维图
ax2 = fig.add_subplot(122)
ax2.scatter(X_pca[:, 0], X_pca[:, 1], c='g')
ax2.set_title('After PCA')
ax2.set_xlabel('Principal Component 1')
ax2.set_ylabel('Principal Component 2')
ax2.grid(True)
plt.show()
```

```python
n = 1000
# 定义主成分方向向量
pc1 = np.random.normal(0, 1, n)
pc2 = np.random.normal(0, 0.2, n)
# 定义第三成分noise
noise = np.random.normal(0, 0.05, n)

# 构建三个特征
X = np.vstack((pc1 + pc2, pc1 - pc2, pc2 + noise)).T
X.shape
```

```python
# PCA降维
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
X_pca.shape
```

```python
# 可视化
# 原始数据3维图
fig = plt.figure(figsize=(12, 4))
ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(X[:, 0], X[:, 1], X[:, 2], c='b')
ax1.set_title('Before PCA')
ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')
ax1.set_zlabel('Feature 3')

# PCA降维后的2维图
ax2 = fig.add_subplot(122)
ax2.scatter(X_pca[:, 0], X_pca[:, 1], c='b')
ax2.set_title('After PCA')
ax2.set_xlabel('Principal Component 1')
ax2.set_ylabel('Principal Component 2')
ax2.grid(True)
plt.show()
```
