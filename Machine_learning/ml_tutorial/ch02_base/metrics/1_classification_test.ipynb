```python
import pandas as pd
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report       # 分类任务评估报告
import seaborn as sns
```

```python
# 定义类标标签
labels = ["cat", "dog"]
# 定义数据(预测值和真实值)
y_true = ["cat", "cat", "cat", "cat", "cat", "cat", "dog", "dog", "dog", "dog"]
y_pred = ["cat", "cat", "dog", "cat", "cat", "cat", "cat", "cat", "dog", "dog"]
```

```python
# 得到混淆矩阵
matrix = confusion_matrix(y_true, y_pred, labels=labels)
print(matrix)
```

```python
df = pd.DataFrame(matrix, columns=labels, index=labels)
df
```

```python
sns.heatmap(matrix, annot=True, fmt="d", cmap="Blues")
```

```python
accuracy = accuracy_score(y_true, y_pred)
accuracy
```

```python
# 猫为正例
precision = precision_score(y_true, y_pred, pos_label="cat")
recall = recall_score(y_true, y_pred, pos_label="cat")
f1 = f1_score(y_true, y_pred, pos_label="cat")
print(precision, recall, f1)
```

```python
# 狗为正例
precision = precision_score(y_true, y_pred, pos_label="dog")
recall = recall_score(y_true, y_pred, pos_label="dog")
f1 = f1_score(y_true, y_pred, pos_label="dog")
print(precision, recall, f1)
```

```python
report = classification_report(y_true, y_pred, labels=labels, target_names=labels,)
print(report)
```
