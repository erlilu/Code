归一化

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
X = [[2,1], [3,1], [1,4], [2,6]]
```

```python
# 定义归一化类的对象
scaler = MinMaxScaler(feature_range=(-1,1))
X_scaler = scaler.fit_transform(X)
print(X_scaler)
```

标准化

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = np.array(X)

scaler = StandardScaler().fit_transform(X)
print(scaler)
```

```python
# print(np.mean(X[:, 0]))
# print(np.mean(X[:, 1]))
print(np.mean(X, axis=0))       # axis=0 :按照行方向压扁，对每列求
```

```python
print(np.std(X, axis=0))
```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
Y = (X - mean) / std
print(Y)
```
